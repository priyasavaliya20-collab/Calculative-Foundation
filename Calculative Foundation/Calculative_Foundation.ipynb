### 🎯 Q1.Represent each student’s subject scores as a vector.


```python
subjects = ["Math", "Physics", "Chemistry", "English", "Computer"]

scores = df[subjects]
print("Select Subject Score Columns")

student_vectors = scores.to_numpy()

print("Convert Scores into NumPy Array (Vectors)")

print(student_vectors)

print("Print Subject Score Vector of Each Student")

for i in range(len(student_vectors)):

   print(f"Student {i+1} Vector = {student_vectors[i]}")

    
```

> 🧮 **Insight:** Every student is now a **5-dimensional vector** in "subject space" — e.g. Student 3 → `[95, 91, 89, 86, 97]` (a strong, well-rounded performer) vs Student 12 → `[58, 62, 60, 68, 64]` (the weakest profile in the batch). Representing students this way is what lets us later apply norms, angles, projections, and PCA to compare them mathematically instead of just eyeballing marksheets. ✅


### 📏 Q2.Compute:
*  Norm-1 and Norm-2 of vectors.



```python
subjects = ["Math", "Physics", "Chemistry", "English", "Computer"]

print("Select Subject Score Columns")
scores = df[subjects]
print(scores.head())

student_vectors = scores.to_numpy()

print("Convert into Vectors")
student_vectors
print(student_vectors[:5])


print("Norm-1 of Each Student\n")

for i, vector in enumerate(student_vectors):
    norm1 = np.linalg.norm(vector, ord=1)
    print(f"Student {i+1}: {norm1}")


print("Norm-2 of Each Student\n")

for i, vector in enumerate(student_vectors):
    norm2 = np.linalg.norm(vector)
    print(f"Student {i+1}: {norm2:.2f}")


```

> 📏 **Insight:** Norm-2 (Euclidean length) ranges from **139.74 (Student 12)** to **207.59 (Student 17)** — the *highest* and *lowest* overall performers by magnitude. Norm-1 and Norm-2 both track total performance, but Norm-2 penalizes/rewards extremes more (squares before summing), so it's the sharper of the two for spotting standout or struggling students. 🔍
>
> ✅ **Conclusion:** Vector norms give a single number that summarizes "how much a student scored overall" — a quick proxy for total marks, useful for ranking without building a full model.

*  🎯 Dot product and angle between two students’ score vectors.


```python
vector1 = student_vectors[0]
vector2 = student_vectors[1]

print(" Select First Two Students")

print("Student 1 Vector :", vector1)
print("Student 2 Vector :", vector2)

dot_product = np.dot(vector1, vector2)

print("Calculate Dot Product")

print("Dot Product =", dot_product)

# Calculate Angle

cos_theta = np.dot(vector1, vector2) / (np.linalg.norm(vector1) * np.linalg.norm(vector2))

angle = np.degrees(np.arccos(cos_theta))


print("Calculate Angle Between Two Vectors")

print("Angle =", round(angle, 2), "degrees")


```

> 🎯 **Insight:** Dot Product = **31518**, Angle ≈ **4.97°** between Student 1 and Student 2. A near-zero angle means their **score patterns point in almost the same direction** — both are relatively stronger/weaker in the same subjects — even though Student 1 scores higher overall (bigger magnitude). 📐
>
> ✅ **Conclusion:** Small angles between student vectors indicate *similar academic strength patterns*, regardless of who scores higher — this is the same idea behind cosine similarity used in recommender systems.

*  ✖️ Cross product (for 3D selected subjects).

```python

print("Select 3 Subjects")

subjects_3d = ["Math", "Physics", "Chemistry"]

scores_3d = df[subjects_3d]

print(scores_3d.head())

print("Convert into 3D Vectors")

vectors_3d = scores_3d.to_numpy()

print(vectors_3d[:2])

print("Select First Two Students")

vector1 = vectors_3d[0]
vector2 = vectors_3d[1]

print("Student 1 :", vector1)
print("Student 2 :", vector2)


print(" Cross Product")

cross_product = np.cross(vector1, vector2)

print("Cross Product =", cross_product)
```

> ✖️ **Insight:** Cross Product = **[203, -392, 168]** for Student 1 vs Student 2 (Math, Physics, Chemistry). Since this vector is **non-zero**, the two students' 3-subject profiles are **not parallel** — they emphasize different subjects to different degrees. Its magnitude also equals the area of the parallelogram the two vectors span in 3D. 🔀
>
> ✅ **Conclusion:** A larger cross-product magnitude would signal more "shape difference" between two students' subject strengths, beyond just their overall level.

### 📐 Q3.Find the projection of one vector onto another.


```python
vector1 = student_vectors[0]
vector2 = student_vectors[1]

print("Select First Two Students")
print("Student 1 Vector :", vector1)
print("Student 2 Vector :", vector2)


projection = (np.dot(vector1, vector2) / np.dot(vector2, vector2)) * vector2

print("Projection of Student 1 onto Student 2 :")
print(projection)

print("Display Rounded Values (Optional)")

print("Projection Vector =", projection.round(2))
```

> 📉 **Insight:** Projecting Student 1 onto Student 2 gives **[81.75, 78.35, 84.02, 91.97, 86.29]** — this is the part of Student 1's performance that "lines up" with Student 2's direction. Notice it pulls Student 1's numbers *down* toward Student 2's scale while keeping the same relative subject pattern. 🎯
>
> ✅ **Conclusion:** Projection is the mathematical base of regression — it answers "if I only had Student 2's pattern to explain Student 1, how much of it would be explained?" 

# 📐 Part B: Matrix Operations

### 📊 Q4. Form a matrix of students × subjects. Perform:


```python

subjects = ["Math", "Physics", "Chemistry", "English", "Computer"]

print("Select Subject Columns")

matrix = df[subjects].to_numpy()

print(matrix)
```

> 🧩 **Insight:** The full **20 × 5 students-by-subjects matrix** is now assembled — 20 rows (students) and 5 columns (Math, Physics, Chemistry, English, Computer). This single matrix is the foundation for every operation that follows: addition, multiplication, transpose, determinant, eigen-decomposition, LU, SVD, PCA. 📊

* ➕ Matrix addition and multiplication.

```python
addition = matrix + matrix

print("Matrix Addition")
print(addition[:5])

multiplication = np.dot(matrix, matrix.T)

print("Matrix Multiplication")
print(multiplication[:5, :5])
```

> ➕ **Insight:** Matrix Addition (A + A) simply **doubles every score** — not analytically interesting on its own, but confirms element-wise addition works correctly. Matrix Multiplication (A · Aᵀ) is more meaningful: its **diagonal entries are the squared L2-norms** of each student (e.g. 36058 = 189.89², matching Student 1's Norm-2 from Q2), and its **off-diagonal entries are dot products between students** (e.g. 31518 between Students 1 & 2, matching Q2's dot product). ✖️
>
> ✅ **Conclusion:** A · Aᵀ is a **similarity (Gram) matrix** — every pairwise relationship between students is packed into one matrix multiplication.

* 🔄 Transpose and Inverse (if possible).

```python
transpose = matrix.T

print("Transpose Matrix")
print(transpose)

square_matrix = matrix[:5, :5]

print("Create Square Matrix")

print(square_matrix)
```

> 🔄 **Insight:** The Transpose flips the 20×5 matrix into 5×20 (subjects as rows, students as columns) — this reshaping is exactly what's needed to compute covariance between *subjects* (Q7) instead of similarity between *students*. A true inverse isn't computed here since the working matrix isn't square (only the 5×5 sub-block from the first 5 students is invertible). 🔁

* 🔢 Determinant.

```python
square_matrix = matrix[:5, :5]

print("Create 5×5 Square Matrix")

print(square_matrix)


determinant = np.linalg.det(square_matrix)

print("Calculate Determinant")

print("Determinant =", determinant)

print("Display Rounded Value (Optional)")
print("Determinant =", round(determinant, 2))
```

> 🔢 **Insight:** Determinant of the 5×5 block (first 5 students) = **−56537**, which is **non-zero**. This confirms the 5 rows are **linearly independent** — none of these 5 students' score patterns can be written as a combination of the others, so the matrix is full rank and safely invertible. ✅
>
> ✅ **Conclusion:** A non-zero determinant is a good sign for any downstream linear algebra (LU, inverse, solving systems) — there's no redundant/degenerate data in this block.

## 🌟 Part D: Eigenvalues & Decomposition


### 🌟 Q7.Compute the eigenvalues and eigenvectors of the covariance matrix. 

```python
subjects = ["Math", "Physics", "Chemistry", "English", "Computer"]

matrix = df[subjects].to_numpy()

print(matrix[:5])


cov_matrix = np.cov(matrix, rowvar=False)

print("Covariance Matrix")
print(cov_matrix)

eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

print("Eigenvalues")
print(eigenvalues)

print("\nEigenvectors")
print(eigenvectors)
```

> 🌟 **Insight:** Eigenvalues of the covariance matrix are **509.96, 8.46, 5.23, 1.81, 0.66**. The **largest eigenvalue alone explains ~96.9% of total variance** among students — meaning almost all the differences between students come down to **one dominant factor: overall academic ability**, rather than 5 independent, unrelated skills. 🔑
>
> ✅ **Conclusion:** This single number is the mathematical proof behind PCA's later success (Q10) — since one direction captures nearly everything, reducing 5 subjects to 1–2 components loses almost no information.


### 🧩 Q8. Perform LU Decomposition of the dataset matrix.  

```python
square_matrix = matrix[:5, :5]

print("Square Matrix")

print(square_matrix)

from scipy.linalg import lu

P, L, U = lu(square_matrix)

print("Permutation Matrix (P)")
print(P)

print("\nLower Triangular Matrix (L)")
print(L)

print("\nUpper Triangular Matrix (U)")
print(U)
```

> 🧩 **Insight:** LU Decomposition splits the matrix into **P** (row-swap/pivoting for numerical stability), **L** (lower-triangular multipliers) and **U** (upper-triangular reduced form). The zero appearing progressively in U's lower-left is exactly what makes solving `Ax = b` or computing the determinant fast — you solve two easy triangular systems instead of one hard one. 🛠️
>
> ✅ **Conclusion:** LU decomposition is the "under the hood" method that libraries like NumPy use whenever you call `det()` or `solve()` — Q4's determinant was effectively computed this way.


### 🧬 Q9. Perform Singular Value Decomposition (SVD) and explain its role in dimensionality reduction. 

```python
U, S, VT = np.linalg.svd(matrix)

print("U Matrix")
print(U)

print("\nSingular Values")
print(S)

print("\nVT Matrix")
print(VT)
```

> 🧬 **Insight:** Singular values are **787.4, 23.4, 10.1, 6.2, 5.3** — the **first singular value alone accounts for ~99.9% of the total "energy"** in the data (since singular values relate to the square root of the eigenvalues seen in Q7, scaled by the uncentered data). This is an even sharper version of the same story: one dominant direction drives almost everything. 🔬
>
> ✅ **Conclusion:** In image compression or recommender systems, this is exactly why keeping just the top 1–2 singular values/vectors reconstructs the data almost perfectly — the same principle applies here to student performance.

## 📉 Part E: Dimensionality Reduction



### 🎯 Q10. Apply Principal Component Analysis (PCA) to reduce the dataset from multiple subjects to 2 dimensions.

```python
# Select Subject Columns

subjects = ["Math", "Physics", "Chemistry", "English", "Computer"]

X = df[subjects]

print(X.head())

# Import PCA

from sklearn.decomposition import PCA

# Reduce to 2 Dimensions

pca = PCA(n_components=2)

pca_data = pca.fit_transform(X)

print(pca_data)

# PCA Result

pca_df = pd.DataFrame(pca_data, columns=["PC1","PC2"])

print(pca_df.head())
```

> 🎯 **Insight:** PCA compresses the 5 subjects into **2 components**. **PC1** ranges from about **−35.8 (Student 12)** to **+32.9 (Student 17)** and clearly tracks *overall performance level* — strong students score high positive PC1, weaker students score large negative PC1. **PC2** has a much smaller spread (**≈ −5 to +4.5**), capturing minor secondary variation (e.g. slightly favoring English/Attendance-style subjects over core sciences). 📉
>
> ✅ **Conclusion:** PC1 essentially re-derives "total marks" from pure geometry, confirming Q7's finding — one dominant academic-ability axis explains almost all the spread between students, and 2D PCA is enough to visualize the entire dataset with barely any information loss.


### 🏷️ Q11. Apply Linear Discriminant Analysis (LDA) to classify students into “Above Average” and “Below Average” categories.

```python
from sklearn.decomposition import PCA

X = df[["Math", "Physics", "Chemistry", "English", "Computer"]].values

pca = PCA(n_components=2)

pca_result = pca.fit_transform(X)

print("PCA Result:")
print(pca_result[:5])
```

> 🏷️ **Insight:** This cell reduces the data with PCA again rather than running a class-supervised LDA — but conceptually, the goal stated in Q11 (separating **"Above Average" vs "Below Average"** students) lines up naturally with PC1: since PC1 already tracks overall ability, students with PC1 > 0 mostly correspond to "Above Average" and PC1 < 0 to "Below Average" in this dataset. 🔖
>
> ✅ **Conclusion:** A true LDA (`sklearn.discriminant_analysis.LinearDiscriminantAnalysis`) would take this further by explicitly maximizing the gap *between* the two result classes rather than just overall variance — worth trying if class separation needs to be sharper than what PCA gives for free.

## ✅ Final Conclusion: Best-Fitting Distribution & Insights

**🏆 Best Fit: Uniform Distribution**

**Why Uniform wins (justification):**

- 📉 **Lowest AIC and BIC** among all five candidates — Uniform explains the data with the fewest parameters and the least model complexity, and still fits as well as or better than the alternatives.
- 📐 **Excess kurtosis ≈ -1.2**, which is almost exactly the theoretical kurtosis of a Uniform distribution (-1.2), versus 0 for Normal. This means the Total Score data is **flatter than a bell curve** — scores are spread fairly evenly across the range rather than clustering around the mean.
- ⚖️ **Skewness ≈ 0** rules out the right-skewed candidates (Log-Normal, Gamma, Exponential), which all assume a long right tail that this data does not have.
- 🎯 **High KS p-value**, meaning we cannot reject the hypothesis that Total Scores come from a Uniform distribution — the observed and theoretical distributions are statistically indistinguishable at conventional significance levels.

**📚 What this means for the dataset:**

- 🎓 Student performance in this batch is **evenly spread across the score range (312–464)** rather than concentrated around a "typical" student — there isn't a strong central tendency pulling most students toward the average.
- 🚫 This is **different from most real-world exam datasets**, which usually follow a roughly Normal (bell-curve) distribution. A Uniform fit here is most likely a **sample-size effect** — with only 20 students, natural randomness can easily produce a flat-looking spread instead of a clear bell shape.
- 📈 **Practical implication:** with such an even spread, simple percentile/rank-based grading (e.g. dividing students into equal-sized performance bands) would be just as fair as z-score-based grading, since there's no dense "middle cluster" for z-scores to model precisely.
- 🔁 **Recommendation:** if this analysis is repeated on a larger cohort (e.g. the 50-student version of this dataset), the Normal distribution is likely to fit better, since Uniform fits are common artifacts of small samples and tend to shift toward Normal as sample size grows. 🌱
